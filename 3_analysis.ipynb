```python
#Advanced Analysis
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
#Correlation Analysis
df = pd.read_csv("cleaned_data.csv")
correlation = df[['price_per_unit', 'quantity', 'revenue', 'age']].corr()
print(correlation)
```

```python
#Customer Segmentation
customer_sales = df.groupby('customer_id')['revenue'].sum()

# High value customers
high_value = customer_sales[customer_sales > customer_sales.mean()]

print("High Value Customers:", len(high_value))
```

```python
# hypothesis testing
from scipy.stats import ttest_ind

male = df[df['gender'] == 'Male']['revenue']
female = df[df['gender'] == 'Female']['revenue']

t_stat, p_value = ttest_ind(male, female)

print("P-value:", p_value)

if p_value < 0.05:
    print("Significant difference between male and female spending")
else:
    print("No significant difference")
```

```python
# Revenue distribution
df['revenue'].hist()
plt.title("Revenue Distribution")
plt.show()
```

```python
# Store type performance
df.groupby('store_type')['revenue'].sum().plot(kind='bar')
plt.title("Store Type Performance")
plt.show()
```
