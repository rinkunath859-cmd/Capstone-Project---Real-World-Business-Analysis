```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned data
df = pd.read_csv("cleaned_data.csv")

# Style
sns.set(style="whitegrid")
```

```python
df.columns
```

```python
#EDA
import matplotlib.pyplot as plt

# Top product categories
top_categories = df.groupby('product_category')['revenue'].sum().sort_values(ascending=False).head(10)

top_categories.plot(kind='bar', title="Top Categories by Revenue")
plt.show()
```

```python
# Sales by city
city_sales = df.groupby('city')['revenue'].sum()

city_sales.plot(kind='pie', autopct='%1.1f%%', title="Sales by City")
plt.show()
```

```python
# Sales trend over time
df.groupby('date')['revenue'].sum().plot(title="Sales Trend Over Time")
plt.show()
```
